```python
#importar o arquivo
import pandas as pd
dados = pd.read_csv(r'..\data\raw\dados_fundo_do_amanha_evcomx.csv', delimiter=';')
display(dados.head())
```

```python
#remover as corridas duplicadas e deixar apenas a primeira ocorrência
dados = dados.drop_duplicates(subset=['corrida'], keep='first')
display(dados.head())
```

```python
#descobrir os tipos de dados
print(dados.dtypes)
```

```python
#identificar os dados faltantes e substituir pela média (tinham 3 na temperaturamediareal)
dados['temperaturamediareal'] = dados['temperaturamediareal'].fillna(dados['temperaturamediareal'].mean())
print(dados.isnull().sum())
```

```python
dados = dados.drop(columns=['temperaturamediareal', 'velocidadereal', 'velocidadeobjetivada', 'sugestaomodelolegado'])
display(dados.head())
```

```python
# Salvar o dataframe processado
import os

# Criar o diretório se não existir
output_dir = r'C:\Users\Maxter\Desktop\dsfa\newcatboostfa\catboost-temperature-prediction\data\processed'
os.makedirs(output_dir, exist_ok=True)

# Salvar o arquivo CSV
output_path = r'C:\Users\Maxter\Desktop\dsfa\newcatboostfa\catboost-temperature-prediction\data\processed\temperature_data.csv'
dados.to_csv(output_path, index=False, sep=';')

print(f"Arquivo salvo em: {output_path}")
print(f"Shape do dataframe final: {dados.shape}")
```
